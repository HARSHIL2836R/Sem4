```python
print(player_spy_series[player_card_series==2])
# plt.xlim(70,90)
for i in range(2,12):
    arr = player_spy_series[player_card_series==i]
    plt.hist(arr,label=i,bins=100)
    print(f"{i} ka min {arr.min()} aur max {arr.max()}")
plt.legend()
plt.show()
```
